```python
import dpsimpy  
```

```python
print(dir(dpsimpy.emt.ph1))
help(dpsimpy.emt.ph1)
```

```python
# Nodes
gnd = dpsimpy.emt.SimNode.gnd
n0  = dpsimpy.emt.SimNode("n0")
n1  = dpsimpy.emt.SimNode("n1")
n2  = dpsimpy.emt.SimNode("n2")
n3  = dpsimpy.emt.SimNode("n3")

v_s = dpsimpy.emt.ph1.VoltageSource("v_s")
v_s.V_ref = 8.5+0j
r = dpsimpy.emt.ph1.Resistor("r")
r.R = 100
r2 = dpsimpy.emt.ph1.Resistor("r2")
r2.R = 100
l = dpsimpy.emt.ph1.Inductor("l")
l.L = 5
c = dpsimpy.emt.ph1.Capacitor("c")
c.C = 250e-6
t = dpsimpy.emt.ph1.IdealTransformer("t")
t.set_parameters(1)

# Connections 
v_s.connect([gnd, n0])
r.connect([n0, n1])
r2.connect([n0, n1])
t.connect([n1, n2])
l.connect([n2, n3])
c.connect([n3, gnd])
```

```python
sys = dpsimpy.SystemTopology(50, [ n0, n1, n2, n3 ], [ v_s, r, l, c, t, r2 ])
sys
```

```python
sim = dpsimpy.Simulation("Ideal Transformer", loglevel=dpsimpy.LogLevel.debug)
sim.set_system(sys)
sim.set_domain(dpsimpy.Domain.EMT)
sim.set_time_step(0.0001)
sim.set_final_time(10)
sim.do_eigenvalue_extraction(True)

log = dpsimpy.Logger("Ideal Transformer")
for i in range(0, len(sys.nodes)):
    log.log_attribute("v" + str(i), "v", sys.nodes[i])

sim.add_logger(log)
    
sim.run()
```
